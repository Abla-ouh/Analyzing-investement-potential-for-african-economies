```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from mpl_toolkits.mplot3d import Axes3D

# Liste des pays
countries_list = [
    'Algeria', 'Angola', 'Benin', 'Botswana', 'Burkina Faso', 'Burundi', 'Cameroon',
    'Cape Verde', 'Central African Republic', 'Chad', 'Comoros', 'Congo',
    'Democratic Republic of the Congo', 'Djibouti', 'Egypt', 'Equatorial Guinea',
    'Eswatini', 'Ethiopia', 'Gabon', 'Gambia', 'Ghana', 'Guinea', 'Guinea-Bissau',
    'Kenya', 'Lesotho', 'Liberia', 'Libya', 'Madagascar', 'Malawi', 'Mali',
    'Mauritania', 'Mauritius', 'Morocco', 'Mozambique', 'Namibia', 'Niger',
    'Nigeria', 'Rwanda', 'São Tomé and Príncipe', 'Senegal', 'Seychelles',
    'Sierra Leone', 'South Africa', 'Sudan', 'Tanzania', 'Togo', 'Tunisia',
    'Uganda', 'Zambia', 'Zimbabwe'
]

# Charger les fichiers CSV et Excel dans des DataFrames
gdp_df = pd.read_csv('africa_gdp_data_2000_2023.csv')  # CSV
inflation_df = pd.read_csv('Inflation_Cleaned_data.csv')  # CSV
exports_df = pd.read_csv('africa_Exports_data_with_regression.csv')  # CSV
imports_df = pd.read_csv('africa_Imports_data_with_regression.csv')  # CSV
Job_df = pd.read_excel('lab.xlsx')  # Excel
FDI_df = pd.read_excel('fdi.xlsx')  # Excel
stability_df = pd.read_excel('stb_politic.xlsx')  # Excel
urbanisation_df = pd.read_excel('urb1.xlsx')  # Excel
technology_df = pd.read_excel('techno.xlsx')  # Excel
energy_df = pd.read_excel('ene1.xlsx')  # Excel
agriculture_df = pd.read_excel('agri1.xlsx')  # Excel
population_df = pd.read_excel('grow_pop.xlsx')  # Excel

# Fonction pour uniformiser le nom de la colonne "Year"
def normalize_year_column(df):
    for col in df.columns:
        if col.lower() in ['year', 'année']:
            df.rename(columns={col: 'Year'}, inplace=True)
            break
    return df

# Normalisation des colonnes "Year" pour tous les fichiers
all_dfs = [gdp_df, inflation_df, exports_df, imports_df, Job_df, FDI_df,
           stability_df, urbanisation_df, technology_df, energy_df, agriculture_df, population_df]

all_dfs = [normalize_year_column(df) for df in all_dfs]

# Fonction pour trouver les pays communs
def find_common_countries(dataframes, countries_list):
    common_countries = set(countries_list)
    for df in dataframes:
        available_countries = set(df.columns) - {'Year'}
        common_countries &= available_countries
    return sorted(common_countries)

# Trouver les pays communs entre toutes les bases
common_countries = find_common_countries(all_dfs, countries_list)

# Fonction pour charger et préparer les données pour une année cible
def load_and_prepare_data(df, countries, target_year):
    df_filtered = df[df['Year'] == target_year]
    return df_filtered[countries].iloc[0].values

# Année cible pour l'analyse
target_year = 2021

# Créer un DataFrame combiné pour l'année cible avec toutes les bases
data_combined = {}
for name, df in zip(
    ['GDP', 'Inflation', 'Exports', 'Imports', 'Job', 'FDI', 'Stability',
     'Urbanisation', 'Technology', 'Energy', 'Agriculture', 'Population'],
    all_dfs
):
    data_combined[name] = load_and_prepare_data(df, common_countries, target_year)

# Conversion en DataFrame
data_2021 = pd.DataFrame(data_combined)

# Normalisation des données
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data_2021)

# Appliquer K-Means
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(scaled_data)

# Ajouter les labels (clusters) au DataFrame
data_2021['Cluster'] = kmeans.labels_


# Choisir trois paramètres pour la visualisation
x_param = 'GDP'
y_param = 'Inflation'
z_param = 'Population'  # Exemple de troisième paramètre

# Créer un graphique 3D des clusters
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Ajouter les points dans l'espace 3D
scatter = ax.scatter(
    data_2021[x_param], data_2021[y_param], data_2021[z_param], 
    c=data_2021['Cluster'], cmap='viridis', s=100, edgecolors='k', alpha=0.7
)

# Ajouter des labels et un titre
ax.set_title(f'Clustering K-Means des pays (avec {x_param}, {y_param}, et {z_param})', fontsize=16)
ax.set_xlabel(x_param, fontsize=14)
ax.set_ylabel(y_param, fontsize=14)
ax.set_zlabel(z_param, fontsize=14)

# Ajouter la barre de couleur pour les clusters
cbar = plt.colorbar(scatter)
cbar.set_label('Cluster', fontsize=14)

# Afficher le graphique
plt.show()
```

```python
sse = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    sse.append(kmeans.inertia_)

plt.plot(range(1, 11), sse)
plt.xlabel('Nombre de clusters')
plt.ylabel('SSE')
plt.title('Méthode du coude')
plt.show()
```

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Suppose data_2021 is the dataframe that contains the correct number of rows
# And you are using this for scaling and clustering
# (Make sure 'data_2021' contains the 39 rows for which you are clustering)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data_2021)  # Scaled data for clustering

# PCA for dimensionality reduction
pca = PCA(n_components=6)
data_reduced = pca.fit_transform(scaled_data)

# K-means clustering with 4 clusters
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(data_reduced)

# Define colors for each cluster (update for 4 clusters)
cluster_colors = ['red', 'green', 'blue', 'purple']  

# Visualize the clustering results
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Assign colors based on the labels
point_colors = [cluster_colors[label] for label in labels]

# Scatter plot with colors corresponding to the clusters
scatter = ax.scatter(
    data_reduced[:, 0], data_reduced[:, 1], data_reduced[:, 2],
    c=point_colors, s=50
)

# Adding labels and axes titles
ax.set_title("K-means clustering with PCA (6 components)")
ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

# Adding legend with labels for clusters (updated for 4 clusters)
ax.legend(
    handles=[
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='red', markersize=10, label="Cluster 1"),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='green', markersize=10, label="Cluster 2"),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='blue', markersize=10, label="Cluster 3"),
        plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='purple', markersize=10, label="Cluster 4"),
    ],
    loc="best"
)

# Show the plot
plt.show()

# Verify that data_2021 and labels have the same number of rows
print(f"Shape of data_2021: {data_2021.shape}")
print(f"Length of labels: {len(labels)}")

# Add the cluster labels to data_2021
data_2021['Cluster'] = labels

# Save the results
data_2021.to_csv("resultat_clustering.csv", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pmdarima.arima import ADFTest
import seaborn as sns
import numpy as np

# Load the data
agriculture_df = pd.read_excel('agri1.xlsx')
energy_df = pd.read_excel('ene1.xlsx')
technology_df = pd.read_excel('techno.xlsx')

# Normalize 'Year' column to 'Year' if necessary
def normalize_year_column(df):
    for col in df.columns:
        if col.lower() in ['year', 'année']:
            df.rename(columns={col: 'Year'}, inplace=True)
            break
    return df

agriculture_df = normalize_year_column(agriculture_df)
energy_df = normalize_year_column(energy_df)
technology_df = normalize_year_column(technology_df)

# List of countries
countries_list = [
    'Algeria', 'Angola', 'Benin', 'Botswana', 'Burkina Faso', 'Burundi', 'Cameroon',
    'Cape Verde', 'Central African Republic', 'Chad', 'Comoros', 'Congo',
    'Democratic Republic of the Congo', 'Djibouti', 'Egypt', 'Equatorial Guinea',
    'Eswatini', 'Ethiopia', 'Gabon', 'Gambia', 'Ghana', 'Guinea', 'Guinea-Bissau',
    'Kenya', 'Lesotho', 'Liberia', 'Libya', 'Madagascar', 'Malawi', 'Mali',
    'Mauritania', 'Mauritius', 'Morocco', 'Mozambique', 'Namibia', 'Niger',
    'Nigeria', 'Rwanda', 'São Tomé and Príncipe', 'Senegal', 'Seychelles',
    'Sierra Leone', 'South Africa', 'Sudan', 'Tanzania', 'Togo', 'Tunisia',
    'Uganda', 'Zambia', 'Zimbabwe'
]

# Function to perform ADF test for a country in a given sector
def perform_adf_test(df, country):
    # Check if the country exists in the dataframe columns
    if country not in df.columns:
        return None  # Return None if the country is not found in the dataset
    
    adf_test = ADFTest(alpha=0.05)
    country_data = df[['Year', country]].dropna()
    country_data.set_index('Year', inplace=True)
    
    # Perform the test on the data
    p_value = adf_test.should_diff(country_data)
    return p_value[0]  # Return the p-value

# Initialize a dictionary to store results
adf_results = {
    'Country': [],
    'Agriculture_P_Value': [],
    'Energy_P_Value': [],
    'Technology_P_Value': []
}

# Loop over each country and perform ADF test for each sector
for country in countries_list:
    # Perform ADF test for Agriculture sector
    agriculture_p_value = perform_adf_test(agriculture_df, country)
    if agriculture_p_value is None:
        agriculture_p_value = 'N/A'  # If the country does not exist in the dataset
    
    # Perform ADF test for Energy sector
    energy_p_value = perform_adf_test(energy_df, country)
    if energy_p_value is None:
        energy_p_value = 'N/A'  # If the country does not exist in the dataset
    
    # Perform ADF test for Technology sector
    technology_p_value = perform_adf_test(technology_df, country)
    if technology_p_value is None:
        technology_p_value = 'N/A'  # If the country does not exist in the dataset
    
    # Append the results to the dictionary
    adf_results['Country'].append(country)
    adf_results['Agriculture_P_Value'].append(agriculture_p_value)
    adf_results['Energy_P_Value'].append(energy_p_value)
    adf_results['Technology_P_Value'].append(technology_p_value)

# Convert results to DataFrame
adf_results_df = pd.DataFrame(adf_results)

# Plotting: Check the stationarity status of the sectors
plt.figure(figsize=(12, 8))

# Replace 'N/A' with NaN for plotting
adf_results_df['Agriculture_P_Value'] = adf_results_df['Agriculture_P_Value'].replace('N/A', np.nan)
adf_results_df['Energy_P_Value'] = adf_results_df['Energy_P_Value'].replace('N/A', np.nan)
adf_results_df['Technology_P_Value'] = adf_results_df['Technology_P_Value'].replace('N/A', np.nan)

# Plot for Agriculture sector
plt.subplot(3, 1, 1)
sns.histplot(adf_results_df['Agriculture_P_Value'].dropna(), kde=True, color='blue')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Agriculture Sector')

# Display p-values below 0.05 on the graph
agriculture_below_05 = adf_results_df['Agriculture_P_Value'][adf_results_df['Agriculture_P_Value'] < 0.05]
for p_val in agriculture_below_05:
    plt.text(p_val, 0.02, f'{p_val:.4f}', color='black', fontsize=8)  # Adjust vertical positioning

plt.legend()

# Plot for Energy sector
plt.subplot(3, 1, 2)
sns.histplot(adf_results_df['Energy_P_Value'].dropna(), kde=True, color='orange')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Energy Sector')

# Display p-values below 0.05 on the graph
energy_below_05 = adf_results_df['Energy_P_Value'][adf_results_df['Energy_P_Value'] < 0.05]
for p_val in energy_below_05:
    plt.text(p_val, 0.02, f'{p_val:.4f}', color='black', fontsize=8)

plt.legend()

# Plot for Technology sector
plt.subplot(3, 1, 3)
sns.histplot(adf_results_df['Technology_P_Value'].dropna(), kde=True, color='red')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Technology Sector')

# Display p-values below 0.05 on the graph
technology_below_05 = adf_results_df['Technology_P_Value'][adf_results_df['Technology_P_Value'] < 0.05]
for p_val in technology_below_05:
    plt.text(p_val, 0.02, f'{p_val:.4f}', color='black', fontsize=8)

plt.legend()

# Adjust layout and show plot
plt.tight_layout()
plt.show()

# Print p-values for each sector below 0.05 in the console
print(f'Agriculture - p-values below 0.05: {agriculture_below_05}')
print(f'Energy - p-values below 0.05: {energy_below_05}')
print(f'Technology - p-values below 0.05: {technology_below_05}')
```

```python
# Function to perform ADF test on differenced series
def perform_adf_test_diff(df, country, d=1):
    # Check if the country exists in the dataframe columns
    if country not in df.columns:
        return None  # Return None if the country is not found in the dataset
    
    adf_test = ADFTest(alpha=0.05)
    country_data = df[['Year', country]].dropna()
    country_data.set_index('Year', inplace=True)
    
    # Apply differencing
    country_data_diff = country_data.diff(periods=d).dropna()  # Differentiation
    
    # Perform the test on the differenced data
    p_value = adf_test.should_diff(country_data_diff)
    return p_value[0]  # Return the p-value

# Initialize a dictionary to store results for differenced series
adf_results_diff = {
    'Country': [],
    'Agriculture_P_Value_d1': [],
    'Energy_P_Value_d1': [],
    'Technology_P_Value_d1': [],
    'Agriculture_P_Value_d2': [],
    'Energy_P_Value_d2': [],
    'Technology_P_Value_d2': []
}

# Loop over each country and perform ADF test on differenced data for each sector (d=1 and d=2)
for country in countries_list:
    # Perform ADF test for Agriculture sector (d=1 and d=2)
    agriculture_p_value_d1 = perform_adf_test_diff(agriculture_df, country, d=1)
    agriculture_p_value_d2 = perform_adf_test_diff(agriculture_df, country, d=2)
    
    # Perform ADF test for Energy sector (d=1 and d=2)
    energy_p_value_d1 = perform_adf_test_diff(energy_df, country, d=1)
    energy_p_value_d2 = perform_adf_test_diff(energy_df, country, d=2)
    
    # Perform ADF test for Technology sector (d=1 and d=2)
    technology_p_value_d1 = perform_adf_test_diff(technology_df, country, d=1)
    technology_p_value_d2 = perform_adf_test_diff(technology_df, country, d=2)
    
    # Append the results to the dictionary
    adf_results_diff['Country'].append(country)
    adf_results_diff['Agriculture_P_Value_d1'].append(agriculture_p_value_d1)
    adf_results_diff['Energy_P_Value_d1'].append(energy_p_value_d1)
    adf_results_diff['Technology_P_Value_d1'].append(technology_p_value_d1)
    adf_results_diff['Agriculture_P_Value_d2'].append(agriculture_p_value_d2)
    adf_results_diff['Energy_P_Value_d2'].append(energy_p_value_d2)
    adf_results_diff['Technology_P_Value_d2'].append(technology_p_value_d2)

# Convert results to DataFrame
adf_results_diff_df = pd.DataFrame(adf_results_diff)

# Print out the results to compare
print(adf_results_diff_df)

# Optionally, you can plot the histograms again for differenced data
plt.figure(figsize=(12, 8))

# Plot for Agriculture sector (d=1 and d=2)
plt.subplot(3, 1, 1)
sns.histplot(adf_results_diff_df['Agriculture_P_Value_d1'].dropna(), kde=True, color='blue', label='d=1')
sns.histplot(adf_results_diff_df['Agriculture_P_Value_d2'].dropna(), kde=True, color='green', label='d=2')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Agriculture Sector (Differenced)')
plt.legend()

# Plot for Energy sector (d=1 and d=2)
plt.subplot(3, 1, 2)
sns.histplot(adf_results_diff_df['Energy_P_Value_d1'].dropna(), kde=True, color='orange', label='d=1')
sns.histplot(adf_results_diff_df['Energy_P_Value_d2'].dropna(), kde=True, color='yellow', label='d=2')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Energy Sector (Differenced)')
plt.legend()

# Plot for Technology sector (d=1 and d=2)
plt.subplot(3, 1, 3)
sns.histplot(adf_results_diff_df['Technology_P_Value_d1'].dropna(), kde=True, color='red', label='d=1')
sns.histplot(adf_results_diff_df['Technology_P_Value_d2'].dropna(), kde=True, color='purple', label='d=2')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Technology Sector (Differenced)')
plt.legend()

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
# Function to perform ADF test after subtracting the moving average
def perform_adf_test_moving_avg(df, country, window=5):
    # Check if the country exists in the dataframe columns
    if country not in df.columns:
        return None  # Return None if the country is not found in the dataset
    
    adf_test = ADFTest(alpha=0.05)
    country_data = df[['Year', country]].dropna()
    country_data.set_index('Year', inplace=True)
    
    # Calculate moving average
    moving_avg = country_data[country].rolling(window=window).mean()
    
    # Subtract the moving average from the original data
    country_data_transformed = country_data[country] - moving_avg
    country_data_transformed = country_data_transformed.dropna()  # Drop NaN values caused by rolling window
    
    # Perform the test on the transformed data
    p_value = adf_test.should_diff(country_data_transformed)
    return p_value[0]  # Return the p-value

# Initialize a dictionary to store results for moving average transformation
adf_results_moving_avg = {
    'Country': [],
    'Agriculture_P_Value_MA': [],
    'Energy_P_Value_MA': [],
    'Technology_P_Value_MA': []
}

# Loop over each country and perform ADF test on transformed data using moving average
for country in countries_list:
    # Perform ADF test for Agriculture sector (using moving average)
    agriculture_p_value_ma = perform_adf_test_moving_avg(agriculture_df, country, window=5)
    
    # Perform ADF test for Energy sector (using moving average)
    energy_p_value_ma = perform_adf_test_moving_avg(energy_df, country, window=5)
    
    # Perform ADF test for Technology sector (using moving average)
    technology_p_value_ma = perform_adf_test_moving_avg(technology_df, country, window=5)
    
    # Append the results to the dictionary
    adf_results_moving_avg['Country'].append(country)
    adf_results_moving_avg['Agriculture_P_Value_MA'].append(agriculture_p_value_ma)
    adf_results_moving_avg['Energy_P_Value_MA'].append(energy_p_value_ma)
    adf_results_moving_avg['Technology_P_Value_MA'].append(technology_p_value_ma)

# Convert results to DataFrame
adf_results_moving_avg_df = pd.DataFrame(adf_results_moving_avg)

# Print out the results to compare
print(adf_results_moving_avg_df)

# Optionally, you can plot the histograms again for the transformed data
plt.figure(figsize=(12, 8))

# Plot for Agriculture sector (moving average transformation)
plt.subplot(3, 1, 1)
sns.histplot(adf_results_moving_avg_df['Agriculture_P_Value_MA'].dropna(), kde=True, color='blue')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Agriculture Sector (Moving Average Transformed)')
plt.legend()

# Plot for Energy sector (moving average transformation)
plt.subplot(3, 1, 2)
sns.histplot(adf_results_moving_avg_df['Energy_P_Value_MA'].dropna(), kde=True, color='orange')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Energy Sector (Moving Average Transformed)')
plt.legend()

# Plot for Technology sector (moving average transformation)
plt.subplot(3, 1, 3)
sns.histplot(adf_results_moving_avg_df['Technology_P_Value_MA'].dropna(), kde=True, color='red')
plt.axvline(x=0.05, color='red', linestyle='--', label='Significance Level (0.05)')
plt.title('ADF Test - Technology Sector (Moving Average Transformed)')
plt.legend()

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pmdarima import auto_arima
from sklearn.metrics import mean_squared_error

def normalize_year_column(df):
    for col in df.columns:
        if col.lower() in ['year', 'année']:
            df.rename(columns={col: 'Year'}, inplace=True)
            break
    return df

def forecast_sector_data(sector):
    # Load the corresponding dataset based on the selected sector
    if sector == 'Agriculture':
        df = pd.read_excel('agri1.xlsx')
    elif sector == 'Technology':
        df = pd.read_excel('techno.xlsx')
    elif sector == 'Energy':
        df = pd.read_excel('ene1.xlsx')
    else:
        raise ValueError("Invalid sector selected. Please choose from 'Agriculture', 'Technology', or 'Energy'.")

    # Normalize the 'Year' column
    df = normalize_year_column(df)

    # List of countries
    countries_list = [
        'Algeria', 'Angola', 'Benin', 'Botswana', 'Burkina Faso', 'Burundi', 'Cameroon',
        'Cape Verde', 'Central African Republic', 'Chad', 'Comoros', 'Congo',
        'Democratic Republic of the Congo', 'Djibouti', 'Egypt', 'Equatorial Guinea',
        'Eswatini', 'Ethiopia', 'Gabon', 'Gambia', 'Ghana', 'Guinea', 'Guinea-Bissau',
        'Kenya', 'Lesotho', 'Liberia', 'Libya', 'Madagascar', 'Malawi', 'Mali',
        'Mauritania', 'Mauritius', 'Morocco', 'Mozambique', 'Namibia', 'Niger',
        'Nigeria', 'Rwanda', 'São Tomé and Príncipe', 'Senegal', 'Seychelles',
        'Sierra Leone', 'South Africa', 'Sudan', 'Tanzania', 'Togo', 'Tunisia',
        'Uganda', 'Zambia', 'Zimbabwe'
    ]

    # Ask user to select a country
    selected_country = input(f"Enter the country from the list {countries_list}: ").capitalize()

    # Check if the selected country exists in the data
    if selected_country not in df.columns:
        raise ValueError(f"The selected country '{selected_country}' is not in the dataset.")
    
    # Extract the data for the selected country
    country_data = df[['Year', selected_country]].dropna()

    # Set 'Year' as the index
    country_data.set_index('Year', inplace=True)
    country_data = country_data.rename(columns={country_data.columns[0]: selected_country})

    # Split data into train (80%) and test (20%)
    train_size = int(len(country_data) * 0.8)
    train_data = country_data.iloc[:train_size]
    test_data = country_data.iloc[train_size:]

    # Fit ARIMA on training data
    model = auto_arima(
        train_data[selected_country], 
        seasonal=False, 
        d=1,
        trace=True,
        suppress_warnings=True,
        stepwise=True
    )

    # Print model summary
    print(model.summary())

    # Make predictions on the test set
    test_predictions = model.predict(n_periods=len(test_data))

    # Evaluate the model
    mse = mean_squared_error(test_data[selected_country], test_predictions)
    print(f"Mean Squared Error on Test Set: {mse}")

    # Forecast future values (2024, 2025)
    future_forecast = model.predict(n_periods=2)
    future_years = [country_data.index.max() + i for i in range(1, 3)]

    # Combine actual data (up to 2023) and forecasted data (2024, 2025)
    combined_data = np.concatenate([
        country_data[selected_country].values,  # Actual data (up to the last year)
        future_forecast                   # Forecast for 2024 and 2025
    ])

    # Combine the corresponding years (actual years + future years)
    combined_years = np.concatenate([
        country_data.index.values,        # Actual years (up to the last year)
        future_years                       # Future years (2024 and 2025)
    ])
    
    # Plot the actual data up to the last year of your dataset and forecast for future years
    plt.figure(figsize=(12, 6))
    plt.plot(combined_years, combined_data, label=f'{selected_country} {sector} Forecast', linestyle='-', marker='o', color='blue')

    # Labels and aesthetics
    plt.xlabel('Year')
    plt.ylabel(f'{selected_country} {sector} Value')
    plt.title(f'{sector} Forecast for {selected_country} - Actual Data + Forecast')
    plt.legend()
    plt.grid()
    plt.show()

    return combined_data, combined_years

# Function usage: Request user input
sector = input("Enter the sector (Agriculture, Technology, Energy): ").capitalize()

# Call the function with the selected sector
forecast_sector_data(sector)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LassoCV, RidgeCV
import matplotlib.pyplot as plt
import seaborn as sns

def create_country_dataset(country, dataframes, start_year=2000):
    """
    Create a consolidated dataset for a specific country with all indicators.
    """
    # Initialize DataFrame with years
    country_data = pd.DataFrame({'Year': range(start_year, 2024)})
    
    # Dictionary to map dataframes to indicator names
    indicator_names = {
        0: 'GDP',
        1: 'Inflation',
        2: 'Exports',
        3: 'Imports',
        4: 'Employment',
        5: 'FDI',
        6: 'Political_Stability',
        7: 'Urbanization',
        8: 'Technology',
        9: 'Energy',
        10: 'Agriculture',
        11: 'Population'
    }
    
    # Add each indicator as a column
    for idx, df in enumerate(dataframes):
        if country in df.columns:
            temp_data = df[['Year', country]].copy()
            temp_data.rename(columns={country: indicator_names[idx]}, inplace=True)
            country_data = country_data.merge(temp_data, on='Year', how='left')
    
    return country_data

def prepare_data_for_modeling(df, target_col='FDI', test_size=0.2):
    """
    Prepare data for modeling by handling missing values and scaling.
    Uses time-based split instead of random split.
    """
    # Sort by year to ensure proper time series split
    df = df.sort_values('Year')
    
    # Calculate split point
    split_idx = int(len(df) * (1 - test_size))
    
    # Separate features and target
    X = df.drop(['Year', target_col], axis=1)
    y = df[target_col]
    
    # Split the data temporally
    X_train = X.iloc[:split_idx]
    X_test = X.iloc[split_idx:]
    y_train = y.iloc[:split_idx]
    y_test = y.iloc[split_idx:]
    
    # Handle missing values
    imputer = SimpleImputer(strategy='mean')
    X_train_imputed = pd.DataFrame(
        imputer.fit_transform(X_train),
        columns=X_train.columns,
        index=X_train.index
    )
    X_test_imputed = pd.DataFrame(
        imputer.transform(X_test),
        columns=X_test.columns,
        index=X_test.index
    )
    
    # Handle missing values in target variable
    y_train = y_train.fillna(y_train.mean())
    y_test = y_test.fillna(y_train.mean())  # Use training mean for test set
    
    # Scale the features
    scaler = StandardScaler()
    X_train_scaled = pd.DataFrame(
        scaler.fit_transform(X_train_imputed),
        columns=X_train.columns,
        index=X_train.index
    )
    X_test_scaled = pd.DataFrame(
        scaler.transform(X_test_imputed),
        columns=X_test.columns,
        index=X_test.index
    )
    
    return X_train_scaled, X_test_scaled, y_train, y_test, scaler, imputer

def predict_future_fdi(country, best_model, scaler, imputer, latest_data, years_to_predict=1):
    """
    Predict FDI for the next specified number of years with rolling updates.
    """
    future_predictions = []
    current_data = latest_data.copy()
    
    for i in range(years_to_predict):
        # Get the most recent year's data
        latest_features = current_data.iloc[-1:].drop(['Year', 'FDI'], axis=1)
        
        # Impute missing values
        latest_features_imputed = pd.DataFrame(
            imputer.transform(latest_features),
            columns=latest_features.columns
        )
        
        # Scale the features
        latest_features_scaled = pd.DataFrame(
            scaler.transform(latest_features_imputed),
            columns=latest_features.columns
        )
        
        # Make prediction for next year
        prediction = best_model.predict(latest_features_scaled)[0]
        
        # Create next year's data
        next_year = current_data.iloc[-1:].copy()
        next_year['Year'] = next_year['Year'] + 1
        next_year['FDI'] = prediction
        
        #  Example of updates of features that typically change year over year because we do not have 2024 data
        # This is a simplified example 
        if 'GDP' in next_year.columns:
            next_year['GDP'] *= 1.02  
        if 'Population' in next_year.columns:
            next_year['Population'] *= 1.01 
        
        # Append to predictions
        future_predictions.append({
            'Country': country,
            'Year': next_year['Year'].iloc[0],
            'Predicted_FDI': prediction
        })
        
        # Add the new row to current_data for next iteration
        current_data = pd.concat([current_data, next_year])
    
    return pd.DataFrame(future_predictions)

def train_and_evaluate_models(X_train, X_test, y_train, y_test):
    """
    Train and evaluate models with proper time series cross-validation.
    """
    # Initialize models with adjusted parameters
    models = {
        'Random Forest': RandomForestRegressor(
            n_estimators=100,
            max_depth=None,
            min_samples_split=5,
            random_state=42
        ),
        'SVR': SVR(kernel='rbf', C=1.0),
        'Ridge': RidgeCV(alphas=[0.1, 1.0, 10.0]),
        'Lasso': LassoCV(alphas=[0.1, 1.0, 10.0])
    }
    
    # Use TimeSeriesSplit with appropriate number of splits
    n_splits = min(3, len(X_train) - 2)
    tscv = TimeSeriesSplit(n_splits=n_splits, test_size=2)
    
    results = {}
    trained_models = {}
    
    for name, model in models.items():
        # Train the model
        model.fit(X_train, y_train)
        trained_models[name] = model
        
        # Make predictions
        y_pred = model.predict(X_test)
        
        # Calculate metrics
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_test, y_pred)
        r2 = r2_score(y_test, y_pred)
        
        # Perform time series cross-validation
        cv_scores = []
        for train_idx, val_idx in tscv.split(X_train):
            X_train_cv = X_train.iloc[train_idx]
            y_train_cv = y_train.iloc[train_idx]
            X_val_cv = X_train.iloc[val_idx]
            y_val_cv = y_train.iloc[val_idx]
            
            model.fit(X_train_cv, y_train_cv)
            y_pred_cv = model.predict(X_val_cv)
            cv_scores.append(r2_score(y_val_cv, y_pred_cv))
        
        results[name] = {
            'MSE': mse,
            'RMSE': rmse,
            'MAE': mae,
            'R2': r2,
            'CV_Scores': np.mean(cv_scores),
            'CV_Std': np.std(cv_scores),
            'Test_Predictions': y_pred
        }
    
    return trained_models, results

def create_visualizations(country_data, model_results, future_predictions, actual_last_years):
    """
    Create comprehensive visualizations for the analysis.
    """
    # Create figure with subplots
    fig = plt.figure(figsize=(20, 12))
    gs = fig.add_gridspec(2, 2)
    
    
    # 2. Model Performance Comparison
    ax2 = fig.add_subplot(gs[0, 0])
    model_names = list(model_results.keys())
    r2_scores = [metrics['R2'] for metrics in model_results.values()]
    cv_scores = [metrics['CV_Scores'] for metrics in model_results.values()]
    
    x = np.arange(len(model_names))
    width = 0.35
    ax2.bar(x - width/2, r2_scores, width, label='R² Score')
    ax2.bar(x + width/2, cv_scores, width, label='CV Score')
    ax2.set_xticks(x)
    ax2.set_xticklabels(model_names, rotation=45)
    ax2.set_title('Model Performance Comparison')
    ax2.set_ylabel('Score')
    ax2.legend()
    ax2.grid(True)
    
    # 3. Error Metrics Comparison
    ax3 = fig.add_subplot(gs[0, 1])
    rmse_scores = [metrics['RMSE'] for metrics in model_results.values()]
    mae_scores = [metrics['MAE'] for metrics in model_results.values()]
    
    x = np.arange(len(model_names))
    width = 0.35
    ax3.bar(x - width/2, rmse_scores, width, label='RMSE')
    ax3.bar(x + width/2, mae_scores, width, label='MAE')
    ax3.set_xticks(x)
    ax3.set_xticklabels(model_names, rotation=45)
    ax3.set_title('Error Metrics Comparison')
    ax3.set_ylabel('Error')
    ax3.legend()
    ax3.grid(True)
    
    # 4. Feature Correlations with FDI
    ax4 = fig.add_subplot(gs[1, 0])
    correlation_data = country_data.drop('Year', axis=1).corr()['FDI'].sort_values()
    correlation_data = correlation_data.drop('FDI')
    sns.heatmap(correlation_data.to_frame(), 
                annot=True, 
                cmap='coolwarm', 
                center=0,
                ax=ax4)
    ax4.set_title('Feature Correlations with FDI')
    
    plt.tight_layout()
    return fig

def analyze_country_fdi(country, all_dataframes, end_year=2023):
    """
    Complete analysis pipeline for a country with visualizations.
    """
    if country not in common_countries:
        return f"{country} is not available in all datasets"
    
    # Create country dataset
    country_data = create_country_dataset(country, all_dataframes)
    
    # Filter data up to end_year
    country_data = country_data[country_data['Year'] <= end_year].copy()
    
    # Print the last few years of actual data
    print("\nLast 5 years of actual data:")
    print(country_data[['Year', 'FDI']].tail())
    
    # Prepare data for modeling
    X_train, X_test, y_train, y_test, scaler, imputer = prepare_data_for_modeling(country_data)
    
    # Train and evaluate models
    trained_models, results = train_and_evaluate_models(X_train, X_test, y_train, y_test)
    
    # Find best model based on test set performance
    best_model_name = max(results.items(), key=lambda x: x[1]['R2'])[0]
    best_model = trained_models[best_model_name]
    
    # Make future predictions
    future_predictions = predict_future_fdi(
        country,
        best_model,
        scaler,
        imputer,
        country_data
    )
    
    # Create visualizations
    visualization_plot = create_visualizations(
        country_data,
        results,
        future_predictions,
        y_test
    )
    
    return {
        'country_data': country_data,
        'model_results': results,
        'best_model': best_model_name,
        'future_predictions': future_predictions,
        'actual_last_years': y_test,
        'visualization_plot': visualization_plot
    }
```
