```python
# Define all African countries and the indicator for political stability
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW"
indicators = "PV.EST"  # Political stability indicator

# Construct the API URL
url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{indicators}?format=json&per_page=10000"

# Send a request to get the data
response = requests.get(url)
data = response.json()

# Extract the relevant information
records = []
for entry in data[1]:
    if entry["value"] is not None and int(entry["date"]) >= 2000:  # Filter out null values and years before 2000
        records.append({
            "Year": entry["date"],
            "Country": entry["country"]["value"],
            "Political Stability": entry["value"]
        })

# Convert the data into a DataFrame
stability_df = pd.DataFrame(records)

# Pivot the DataFrame to have the required structure
stability_pivot = stability_df.pivot(index="Year", columns="Country", values="Political Stability")

# Reset index to bring "Year" into the columns
stability_pivot = stability_pivot.reset_index()
```

```python
stability_pivot.head()
#The values in stability_pivot give the country's score on the aggregate indicator, 
#in units of a standard normal distribution
```

```python
#remove Country
stability_pivot.columns.name = None
```

```python
#Since the indicator is in units of a standard normal distribution, 
#we can compute the probability to have the data beetween 0 and 1.
```

```python
from scipy.stats import norm
for col in stability_pivot.columns[1:]:
    stability_pivot[col] = stability_pivot[col].apply(norm.cdf)
```

```python
def replace_missing(df):
    # Iterate over the rows of the DataFrame
    for i in range(len(df)):
        for j in range(df.shape[1]):
            # If the value is missing
            if pd.isna(df.iloc[i, j]):
                if i == len(df) - 1:  # Last row
                    # Replace with the average of the previous two values
                    if j >= 2:
                        df.iloc[i, j] = np.mean([df.iloc[i-1, j], df.iloc[i-2, j]])
                    elif j == 1:
                        df.iloc[i, j] = df.iloc[i-1, j]  # If there is only one value before
                    elif j == 0:
                        continue  # Cannot replace the first value
                else:
                    # Replace with the average of the previous two and next two non-missing values
                    prev_vals = []
                    next_vals = []

                    # Previous values
                    if j > 0 and pd.notna(df.iloc[i, j-1]):
                        prev_vals.append(df.iloc[i, j-1])
                    if j > 1 and pd.notna(df.iloc[i, j-2]):
                        prev_vals.append(df.iloc[i, j-2])
                    # Next values
                    if j < df.shape[1]-1 and pd.notna(df.iloc[i, j+1]):
                        next_vals.append(df.iloc[i, j+1])
                    if j < df.shape[1]-2 and pd.notna(df.iloc[i, j+2]):
                        next_vals.append(df.iloc[i, j+2])

                    # Average of the previous and next non-missing values
                    if prev_vals and next_vals:
                        df.iloc[i, j] = np.mean(prev_vals + next_vals)
                    elif prev_vals:
                        df.iloc[i, j] = np.mean(prev_vals)
                    elif next_vals:
                        df.iloc[i, j] = np.mean(next_vals)

    return df
```

```python
# Fonction pour remplacer les valeurs manquantes par la moyenne des deux qui succedent en vérifiant du bas vers le haut
def replace_missing_values(df):
    for column in df.columns:
        for row in range(len(df) - 2, -1, -1):  # Commence par la dernière ligne et remonte
            if pd.isna(df.iloc[row, df.columns.get_loc(column)]):  # Si la valeur est manquante
                # Remplacer par la moyenne des deux lignes suivantes
                next_value = df.iloc[row + 1, df.columns.get_loc(column)]
                next_next_value = df.iloc[row + 2, df.columns.get_loc(column)] if row + 2 < len(df) else next_value
                df.iloc[row, df.columns.get_loc(column)] = (next_value + next_next_value) / 2
    return df
```

```python
import requests
import pandas as pd

# Define the countries and indicators
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW" 
fdi_indicator = "BX.KLT.DINV.CD.WD"  # FDI Inflows
gdp_indicator = "NY.GDP.MKTP.CD"  # GDP (current US$)

# Fetch FDI data from the World Bank API
fdi_url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{fdi_indicator}?format=json&per_page=10000"
fdi_response = requests.get(fdi_url)
fdi_data = fdi_response.json()

# Extract FDI data
fdi_records = []
for entry in fdi_data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filter for years within the range 2000-2023
        if 2000 <= year <= 2023:
            fdi_records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "FDI": entry["value"]
            })

# Convert FDI data to DataFrame
dfF = pd.DataFrame(fdi_records)

# Fetch GDP data from the World Bank API
gdp_url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{gdp_indicator}?format=json&per_page=10000"
gdp_response = requests.get(gdp_url)
gdp_data = gdp_response.json()

# Extract GDP data
gdp_records = []
for entry in gdp_data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filter for years within the range 2000-2023
        if 2000 <= year <= 2023:
            gdp_records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "GDP": entry["value"]
            })

# Convert GDP data to DataFrame
dfG = pd.DataFrame(gdp_records)

# Merge FDI and GDP data on Country and Year
merged_df = pd.merge(dfF, dfG, on=["Country", "Year"],how="inner")

# Calculate FDI as a percentage of GDP
merged_df["FDI_as_percentage_of_GDP"] = (merged_df["FDI"] / merged_df["GDP"]) * 100

# Reshape data: Rows = Years, Columns = Countries, Values = FDI as % of GDP
pivot_dff = merged_df.pivot(index="Year", columns="Country", values="FDI_as_percentage_of_GDP")

# Sort rows by Year
pivot_dff.sort_index(inplace=True)

# Save to CSV
pivot_dff.to_csv("africa_FDI_as_percentage_of_GDP.csv")
print("Data saved to africa_FDI_as_percentage_of_GDP.csv")

# Display the first few rows
pivot_dff.tail()
```

```python
##charger la base du pib et verifier la correlation pour une  prediction avec le pib
import requests
import pandas as pd

# Définir les pays et l'indicateur PIB
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW"
gdp_indicator = "NY.GDP.MKTP.CD"  # PIB (en USD)

# URL de l'API de la Banque Mondiale pour récupérer les données du PIB
gdp_url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{gdp_indicator}?format=json&per_page=10000"
gdp_response = requests.get(gdp_url)
gdp_data = gdp_response.json()

# Extraire les données du PIB et les diviser par 10^9 (pour les afficher en milliards)
gdp_records = []
for entry in gdp_data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filtrer les années entre 2000 et 2023
        if 2000 <= year <= 2023:
            gdp_records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "GDP": entry["value"] / 1e9  # Diviser par 10^9 pour obtenir des milliards
            })

# Convertir les données du PIB en DataFrame
dfG = pd.DataFrame(gdp_records)

# Pivot: Les années en ligne, les pays en colonne, les valeurs de PIB en milliards
pivot_df = dfG.pivot(index="Year", columns="Country", values="GDP")

# Afficher les premières lignes des données
pivot_df.head()

```

```python
from sklearn.linear_model import LinearRegression
import numpy as np


# Assurez-vous que pivot_dfE et pivot_df sont bien alignés avec les mêmes années et pays
# Pivot_df doit contenir les données du PIB
# Filtrer les pays avec des valeurs manquantes dans pivot_dfE

missing_countries = pivot_dfE.columns[pivot_dfE.isnull().any()]

# Créer un modèle de régression linéaire
model = LinearRegression()

# Parcourir les pays avec des valeurs manquantes
for country in missing_countries:
    # Sélectionner les années avec des données disponibles pour ce pays
    available_data = pivot_dfE[country].dropna()
    years_with_data = available_data.index

    # Vérifier que les années correspondantes existent dans pivot_df
    gdp_data = pivot_df.loc[years_with_data, country].values.reshape(-1, 1)
    export_data = available_data.values

    # Effectuer la régression linéaire entre le PIB et les exportations
    model.fit(gdp_data, export_data)

    # Prévoir les exportations manquantes en utilisant les données du PIB
    missing_years = pivot_dfE.index[pivot_dfE[country].isnull()]
    
    # Assurez-vous que les années manquantes existent aussi dans pivot_df
    gdp_missing = pivot_df.loc[missing_years, country].values.reshape(-1, 1)

    # Prédire les exportations pour les années manquantes
    predicted_exports = model.predict(gdp_missing)
    predicted_exports = np.clip(predicted_exports, 0, None)

    # Remplir les valeurs manquantes dans pivot_dfE avec les valeurs prédites
    pivot_dfE.loc[missing_years, country] = predicted_exports


# Vérifier les valeurs manquantes après la régression
print("\nMissing values after regression:")
print(pivot_dfE.isnull().sum())

# Sauvegarder les données remplies dans un fichier CSV
pivot_dfE.to_csv("africa_Exports_data_with_regression.csv")
#print("Interpolated data using regression saved to africa_Exports_data_interpolated_with_regression.csv")

# Afficher les premières lignes des données remplies
print("\nInterpolated data using regression:")
pivot_dfE.head()

```
