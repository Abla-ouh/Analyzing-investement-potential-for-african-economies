```python
import requests
import pandas as pd

# Define the countries and indicators
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW"
indicator = "FP.CPI.TOTL.ZG"  # inflation 

# Fetch data from the World Bank API
url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{indicator}?format=json&per_page=10000"

response = requests.get(url)
data = response.json()

# Extract relevant information
records = []
for entry in data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filter for years within the range 2000-2023
        if 2000 <= year <= 2023:
            records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "Inflation": entry["value"]
            })

# Convert to DataFrame
dfI = pd.DataFrame(records)

# Reshape data: Rows = Years, Columns = Countries
pivot_dfI = dfI.pivot(index="Year", columns="Country", values="Inflation")

# Sort rows by Year
pivot_dfI.sort_index(inplace=True)

# Convert GDP values to billions for readability
#pivot_df = pivot_df / 1e9

# Save to CSV
pivot_dfI.to_csv("africa_inflation_2000_2023_data.csv")
print("Data saved to africa_inflation_data.csv")

# Display the first few rows
pivot_dfI.tail()
```

```python
from sklearn.linear_model import LinearRegression

# 3. Combine inflation and GDP data into one DataFrame
pivot_df_combined = pivot_dfI.copy()  # Start with inflation data
pivot_df_combined['GDP'] = pivot_df.mean(axis=1)  # Add GDP data (mean across countries for each year)

# 4. Prepare the data for regression (use GDP as predictor, inflation as target)
# Remove rows where GDP or inflation is missing
complete_data = pivot_df_combined.dropna()  # Remove rows with missing values
X = complete_data['GDP'].values.reshape(-1, 1)  # GDP is the feature
y = complete_data.iloc[:, :-1]  # Inflation data is the target (All columns except 'GDP')

# 5. Train the regression model
model = LinearRegression()
model.fit(X, y)

# 6. Predict missing inflation values using the trained model
missing_data = pivot_df_combined[pivot_df_combined.isnull().any(axis=1)]  # Rows with missing values
X_missing = missing_data['GDP'].values.reshape(-1, 1)
predicted_values = model.predict(X_missing)

# 7. Fill the missing inflation values with the predicted values
# Iterate over each column (country) and fill missing values with predicted values
predicted_values_idx = 0  # Index to track the position in predicted values
for col in pivot_df_combined.columns[:-1]:  # Skip the last column (GDP)
    # Identify rows where the current column has missing values
    missing_indices = pivot_df_combined[col].isnull()

    # If there are missing values in this column, fill them
    if missing_indices.any():
        # Make sure we fill only the missing values in the current column
        pivot_df_combined.loc[missing_indices, col] = predicted_values[predicted_values_idx, 0:missing_indices.sum()]
        predicted_values_idx += 1  # Move to the next set of predicted values

# 8. Save the imputed data to a CSV file
pivot_df_combined.drop('GDP', axis=1, inplace=True)
pivot_df_combined.to_csv("africa_inflation_regression_imputed.csv")
print("Imputed data saved to africa_inflation_regression_imputed.csv")

# 9. Display the imputed DataFrame
pivot_df_combined.head()
```

```python
pivot_dfI.interpolate(method='linear', axis=0, inplace=True)
# Fonction pour remplacer les valeurs manquantes par la moyenne des deux qui succedent en vérifiant du bas vers le haut
def replace_missing_values(df):
    for column in df.columns:
        for row in range(len(df) - 2, -1, -1):  # Commence par la dernière ligne et remonte
            if pd.isna(df.iloc[row, df.columns.get_loc(column)]):  # Si la valeur est manquante
                # Remplacer par la moyenne des deux lignes suivantes
                next_value = df.iloc[row + 1, df.columns.get_loc(column)]
                next_next_value = df.iloc[row + 2, df.columns.get_loc(column)] if row + 2 < len(df) else next_value
                df.iloc[row, df.columns.get_loc(column)] = (next_value + next_next_value) / 2
    return df

inflation = replace_missing_values(pivot_dfI)
inflation.to_csv("Inflation_Cleaned_data.csv")
print("Data saved to Inflation_Cleaned_data.csv")
```

```python
import requests
import pandas as pd

# Define the countries and indicators
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW" 
fdi_indicator = "BX.KLT.DINV.CD.WD"  # FDI Inflows
gdp_indicator = "NY.GDP.MKTP.CD"  # GDP (current US$)

# Fetch FDI data from the World Bank API
fdi_url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{fdi_indicator}?format=json&per_page=10000"
fdi_response = requests.get(fdi_url)
fdi_data = fdi_response.json()

# Extract FDI data
fdi_records = []
for entry in fdi_data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filter for years within the range 2000-2023
        if 2000 <= year <= 2023:
            fdi_records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "FDI": entry["value"]
            })

# Convert FDI data to DataFrame
dfF = pd.DataFrame(fdi_records)

# Fetch GDP data from the World Bank API
gdp_url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{gdp_indicator}?format=json&per_page=10000"
gdp_response = requests.get(gdp_url)
gdp_data = gdp_response.json()

# Extract GDP data
gdp_records = []
for entry in gdp_data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filter for years within the range 2000-2023
        if 2000 <= year <= 2023:
            gdp_records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "GDP": entry["value"]
            })

# Convert GDP data to DataFrame
dfG = pd.DataFrame(gdp_records)

# Merge FDI and GDP data on Country and Year
merged_df = pd.merge(dfF, dfG, on=["Country", "Year"],how="inner")

# Calculate FDI as a percentage of GDP
merged_df["FDI_as_percentage_of_GDP"] = (merged_df["FDI"] / merged_df["GDP"]) * 100

# Reshape data: Rows = Years, Columns = Countries, Values = FDI as % of GDP
pivot_dff = merged_df.pivot(index="Year", columns="Country", values="FDI_as_percentage_of_GDP")

# Sort rows by Year
pivot_dff.sort_index(inplace=True)

# Save to CSV
pivot_dff.to_csv("africa_FDI_as_percentage_of_GDP.csv")
print("Data saved to africa_FDI_as_percentage_of_GDP.csv")

# Display the first few rows
pivot_dff.tail()
```

```python
import requests
import pandas as pd

# Define the countries and indicators
countries = "DZ;AO;BJ;BW;BF;BI;CM;CV;CF;TD;KM;CG;CD;DJ;EG;GQ;SZ;ET;GA;GM;GH;GN;GW;KE;LS;LR;LY;MG;MW;ML;MR;MU;MA;MZ;NA;NE;NG;RW;ST;SN;SC;SL;ZA;SD;TZ;TG;TN;UG;ZM;ZW"
indicator = "NE.EXP.GNFS.CD"  # Exports 

# Fetch data from the World Bank API
url = f"http://api.worldbank.org/v2/country/{countries}/indicator/{indicator}?format=json&per_page=10000"

response = requests.get(url)
data = response.json()

# Extract relevant information
records = []
for entry in data[1]:
    if entry["value"]:
        year = int(entry["date"])
        # Filtrer pour les années entre 2000 et 2023
        if 2000 <= year <= 2023:
            records.append({
                "Country": entry["country"]["value"],
                "Year": year,
                "Exports": entry["value"]
            })

# Convert to DataFrame
dfE = pd.DataFrame(records)

# Reshape data: Rows = Years, Columns = Countries
pivot_dfE = dfE.pivot(index="Year", columns="Country", values="Exports")

# Sort rows by Year
pivot_dfE.sort_index(inplace=True)


# Convert  values to billions for readability
pivot_dfE = pivot_dfE / 1e9

# Save to CSV
pivot_dfE.to_csv("africa_Exports_data.csv")
print("Data saved to africa_Exports_data.csv")

# Display the first few rows
pivot_dfE.tail()
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np


# Assurez-vous que pivot_dfE et pivot_df sont bien alignés avec les mêmes années et pays
# Pivot_df doit contenir les données du PIB
# Filtrer les pays avec des valeurs manquantes dans pivot_dfE

missing_countries = pivot_dfE.columns[pivot_dfE.isnull().any()]

# Créer un modèle de régression linéaire
model = LinearRegression()

# Parcourir les pays avec des valeurs manquantes
for country in missing_countries:
    # Sélectionner les années avec des données disponibles pour ce pays
    available_data = pivot_dfE[country].dropna()
    years_with_data = available_data.index

    # Vérifier que les années correspondantes existent dans pivot_df
    gdp_data = pivot_df.loc[years_with_data, country].values.reshape(-1, 1)
    export_data = available_data.values

    # Effectuer la régression linéaire entre le PIB et les exportations
    model.fit(gdp_data, export_data)

    # Prévoir les exportations manquantes en utilisant les données du PIB
    missing_years = pivot_dfE.index[pivot_dfE[country].isnull()]
    
    # Assurez-vous que les années manquantes existent aussi dans pivot_df
    gdp_missing = pivot_df.loc[missing_years, country].values.reshape(-1, 1)

    # Prédire les exportations pour les années manquantes
    predicted_exports = model.predict(gdp_missing)
    predicted_exports = np.clip(predicted_exports, 0, None)

    # Remplir les valeurs manquantes dans pivot_dfE avec les valeurs prédites
    pivot_dfE.loc[missing_years, country] = predicted_exports


# Vérifier les valeurs manquantes après la régression
print("\nMissing values after regression:")
print(pivot_dfE.isnull().sum())

# Sauvegarder les données remplies dans un fichier CSV
pivot_dfE.to_csv("africa_Exports_data_with_regression.csv")
#print("Interpolated data using regression saved to africa_Exports_data_interpolated_with_regression.csv")

# Afficher les premières lignes des données remplies
print("\nInterpolated data using regression:")
pivot_dfE.head()






```
