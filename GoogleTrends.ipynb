```python
# File path: google_trends_political_stability.py

from pytrends.request import TrendReq
import pandas as pd

# Initialize Pytrends
def fetch_google_trends(keywords, country_code, timeframe='today 12-m'):
    """
    Fetches Google Trends data for specific keywords and region.

    :param keywords: List of keywords to query (e.g., ['political stability', 'civil unrest']).
    :param country_code: Geo location code (e.g., 'ZA' for South Africa).
    :param timeframe: Time range for data (default: last 12 months).
    :return: Dataframe with trends data.
    """
    pytrends = TrendReq()
    pytrends.build_payload(keywords, geo=country_code, timeframe=timeframe)
    trends_data = pytrends.interest_over_time()

    if not trends_data.empty:
        trends_data = trends_data.drop(columns=['isPartial'])
    return trends_data

# Analyze Data to Determine Stability
def determine_stability(trends_data, stability_keywords, conflict_keywords):
    """
    Analyzes trends data to determine stability.

    :param trends_data: Dataframe of trends data.
    :param stability_keywords: Keywords indicating stability.
    :param conflict_keywords: Keywords indicating conflict.
    :return: Stability score (-100 to 100).
    """
    trends_data['stability_score'] = trends_data[stability_keywords].mean(axis=1) - trends_data[conflict_keywords].mean(axis=1)
    return trends_data

# Example Usage
if __name__ == "__main__":
    # Define countries and keywords
    countries = {'South Africa': 'ZA', 'Kenya': 'KE', 'Nigeria': 'NG'}
    stability_keywords = ['political stability', 'peaceful elections']
    conflict_keywords = ['civil unrest', 'political conflict', 'protests']

    # Collect and analyze data
    all_data = {}
    for country, code in countries.items():
        print(f"Fetching data for {country}...")
        data = fetch_google_trends(stability_keywords + conflict_keywords, code)
        if not data.empty:
            analyzed_data = determine_stability(data, stability_keywords, conflict_keywords)
            all_data[country] = analyzed_data

            # Save to CSV
            analyzed_data.to_csv(f"{country}_political_stability.csv")
            print(f"Data for {country} saved as {country}_political_stability.csv")
        else:
            print(f"No data found for {country}")

    # Summary
    print("Data collection completed.")
```

```python
from pytrends.request import TrendReq
import pandas as pd
import matplotlib.pyplot as plt

# Function to get trends data for a search term and countries
def get_trends_data(keywords, countries, timeframe='today 12-m'):
    pytrends = TrendReq(hl='en-US', tz=360)
    
    # Create a dictionary to store data
    trends_data = {}
    
    for country in countries:
        print(f"Fetching data for {country}...")
        pytrends.build_payload(keywords, geo=country, timeframe=timeframe)
        data = pytrends.interest_over_time()
        
        # Store the data for each country
        trends_data[country] = data[keywords[0]]  # Store the trend for the first keyword
    
    return trends_data

# Define the search terms
keywords = ['green projects', 'digital transformation']

# List of African countries' codes (ISO-3166-1 alpha-2 codes)
african_countries = [
     'NG'
]

# Get the trends data for green projects and digital transformation
trends_data = get_trends_data(keywords, african_countries)

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(trends_data)

# Plot the trends for each country
plt.figure(figsize=(15, 10))
for country, data in trends_data.items():
    plt.plot(data.index, data, label=country)

plt.title('Google Trends - Green Projects vs Digital Transformation (Over Time)')
plt.xlabel('Date')
plt.ylabel('Interest Over Time')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Show the plot
plt.show()

# Optionally, save the data to CSV for further analysis
df.to_csv('african_countries_trends.csv', index=True)

print("Data saved to 'african_countries_trends.csv'.")
```

```python
from pytrends.request import TrendReq
import pandas as pd
import matplotlib.pyplot as plt

# Connexion à Google Trends
pytrends = TrendReq(hl='en-US', tz=360)

# Mots-clés à analyser
keywords = ['digital transformation']

# Construire le payload pour récupérer les données sur 3 ans
pytrends.build_payload(keywords, timeframe='2024-01-01 2024-01-08', geo='NG')  # Exemple pour les États-Unis

# Récupérer l'intérêt au fil du temps
interest_over_time_df = pytrends.interest_over_time()



# Regrouper les données par mois (en moyennant l'intérêt sur chaque mois)
monthly_data = interest_over_time_df  #.resample('M').mean()

# Afficher les données mensuelles
print(monthly_data)

# Visualisation des tendances mensuelles
plt.figure(figsize=(10, 6))
plt.plot(monthly_data.index, monthly_data['digital transformation'], label='Digital Transformation')
plt.plot(monthly_data.index, monthly_data['green projects'], label='Green Projects')

plt.title('Google Trends for Digital Transformation and Green Projects (Last 3 Years)')
plt.xlabel('Time')
plt.ylabel('Monthly Interest')
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

# Afficher le graphique
plt.show()

```
